<b>Problem 1: A fair coin is tossed 100 times. Calculate:</b>
+ Number of heads
+ Number of tails
+ Experimental probability of head
+ Experimental probability of tails

```python
CoinTossEvents = ['H', 'T']
def FlipACoin(n: int):
    return rng.choice(CoinTossEvents, size = n)
S1 = FlipACoin(100)
S1
```

```python
def HeadsCount(S):
    return S[S == 'H'].size
def TailsCount(S):
    return S[S == 'T'].size
def HeadsProbability(S):
    return HeadsCount(S)/S.size
def TailsProbability(S):
    return TailsCount(S)/S.size
def Problem1Output(S):
    p("Number of heads:", HeadsCount(S))
    p("Number of tails:", TailsCount(S))
    p("Experimental probability of heads:", HeadsProbability(S))
    p("Experimental probability of tails:", TailsProbability(S))
Problem1Output(S1)
```

<b>Problem 3: Rolling a Die</b><br>
A fair die is rolled 6000 times.
Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}

```python
DieRollEvents=[1,2,3,4,5,6]
def RollADie(n: int):
    return rng.choice(DieRollEvents, size = n)
def RollCount(S, n):
    return S[S == n].size
def RollProbability(S, n):
    x=RollCount(S,n)/S.size
    p("Probability of", n, ":", x)
    return x
```

```python
S6=RollADie(6000)
[RollProbability(S6,x) for x in DieRollEvents]
```

<b>Problem 4: Addition rule for disjoint events</b><br>
A die is rolled 10000 times<br>
Define: A=getting 1, B= getting 6, check the events are joint or disjoint<br>
Verify: P(AuB)=P(A)+P(B)

```python
S7=RollADie(10000)
A=RollProbability(S7,1)
B=RollProbability(S7,6)
#A and B are disjoint events as event cannot be both 1 AND 6
AuB=S7[(S7 == 1) | (S7 == 6)].size/S7.size
p("P(AuB):",AuB)
p("P(A) + P(B):",A+B)
```

<b>Problem 5: Addition rule for overlapping events</b><br>
For one die roll, define A = Getting an even number (A = {2,4,6}) and B = getting a number greater than 3 (B = {4,5,6})<br>
Verify: P(AuB) = P(A) + P(B) - P(AnB)

```python
SS={1,2,3,4,5,6}
A={2,4,6}
B={4,5,6}
p("A:",A)
P_A=len(A)/len(SS)
p("\tP(A):",P_A)
p("B:",B)
P_B=len(B)/len(SS)
p("\tP(B):",P_B)
AnB=A.intersection(B)
p("AnB:",AnB)
P_AnB=len(AnB)/len(SS)
p(f"\tP(AnB): {P_AnB:.5f}")
p(f"\nP(A) + P(B) - P(AnB): {(P_A + P_B - P_AnB):.5f}")
AuB=A.union(B)
p("\nAuB:",AuB)
P_AuB=len(AuB)/len(SS)
p(f"\tP(AuB): {P_AuB:.5f}")
p()
```
