# Problem Set 1: Simulating Customer Clicks Using a Bernoulli Random Variable

## Problem Statement

An organization sends promotional emails. Historical data indicate that a customer clicks the link with probability $p = 0.30$. 

Define:
$$X = \begin{cases} 1, & \text{customer clicks} \\ 0, & \text{customer does not click} \end{cases}$$

Thus,
$$X \sim \text{Bernoulli}(0.30)$$

---

## Tasks

### Step 1: Identify the Random Variable
Answer the following:
* What does $X = 1$ represent?<br>
>>>It represents the event that the customer clicks the link
* What does $X = 0$ represent?
>>>It represents the event that the customer does not click the link
* What are the possible values of $X$?
>>>The possible values are 1 or 0
* Is $X$ a discrete or continuous random variable?
>>>Discrete
* Which probability distribution is suitable for $X$?
>>>Bernoulli Distribution

---

### Step 2: Write the Theoretical PMF
Write down the theoretical Probability Mass Function (PMF) for the random variable $X$.

---
$$P(X = x) = p^x (1 - p)^{1 - x} \quad \text{for } x \in \{0, 1\}$$


### Step 3: Generate Outcomes for 20 Customers
*Hint: `np.random.binomial(n=1, ...)` generates Bernoulli observations because only one trial is performed for each customer.*

Use code:
```python
samples = np.random.binomial(n=1, p=p, size=sample_size)
```



````python
import numpy as np
p = 0.3
def roll(prob, sample_size, num = 1):
    samples = np.random.binomial(num,prob, sample_size)
    return samples
sample10 =  roll(0.3, 20)
print("10 rolls:", sample10)
````

Answer the following:
* How many observations were generated?
>>>10
* What does each zero represent?
>>>Event that customer did not click the link
* What does each one represent?
>>>Event that customer clicked the link
* Are your generated observations the same as the theoretical probabilities?
>>>No, slight variations

### Step 7: Plot Empirical and Theoretical PMFs
Create a visualization comparing the theoretical PMF with your empirical PMF derived from the simulation.

---



````python
import matplotlib.pyplot as plt
import numpy as np

x_vals = [0, 1]
theoretical_pmf = [0.70, 0.30]
empirical_pmf = [number_of_non_clicks / 20, number_of_clicks / 20]

x = np.arange(len(x_vals))
width = 0.3

plt.bar(x - width/2, theoretical_pmf, width, label='Theoretical', color='skyblue')
plt.bar(x + width/2, empirical_pmf, width, label='Empirical', color='orange')
plt.xticks(x, x_vals)
plt.xlabel('Outcome')
plt.ylabel('Probability')
plt.title('Theoretical vs. Empirical PMF (n = 20)')
plt.legend()
plt.show()
````

### Step 8: Calculate the Theoretical Expectation
Compute $E[X]$ using the theoretical parameter $p$.

---
$$E(X)=\sum_{i=1}^{20} x_{i}P(x_{i})$$
$$E(X)= (1*0.3) + (0 * 0.7)$$
$$E(X) = 0.30$$


### Step 9: Calculate the Empirical Expectation
Compute the sample mean from your generated observations.

---
$$E(X)=\sum_{i=1}^{20} x_{i}P(x_{i})$$
$$E(X)= (1*0.2) + (0 * 0.8)$$
$$E(X) = 0.20$$


### Step 10: Calculate the Theoretical Variance
Compute $\text{Var}(X)$ using the theoretical parameter $p$.

---
 $$Var(X) = E[X^2] - (E[X])^2$$
 $$Var(X) = 0.3 - (0.3)^2$$
 $$Var(X) = 0.3 - 0.09$$
 $$Var(X) = 0.21$$


### Step 11: Calculate the Empirical Variance
Compute the sample variance from your generated observations.

---
 $$Var(X) = E[X^2] - (E[X])^2$$
 $$Var(X) = 0.2 - (0.2)^2$$
 $$Var(X) = 0.2 - 0.04$$
 $$Var(X) = 0.16$$

# Problem Set 2: ML Use Case — Number of Fraudulent Transactions Flagged

## Problem Statement

A fraud-detection model examines $50$ transactions every hour. Assume that each transaction has a $4\%$ probability of being flagged as potentially fraudulent.

Let:

* $X =$ number of transactions flagged in one hour

Initially assume that individual flagging decisions are independent:

$$X \sim \text{Binomial}(n = 50, p = 0.04)$$

The fraud investigation team can examine at most four flagged transactions per hour.

---

## Tasks

### Task 1

Simulate the operation of the fraud-detection system for **20,000 independent one-hour periods**.

During every simulated hour:

* Exactly $50$ transactions are examined.
* Each transaction has a flagging probability of $0.04$.
* $X$ records the total number of flagged transactions.

> **Hint:** Use the `np.random.binomial` function in Python.

---


````python
import numpy as np
np.random.seed(2002)
number_of_transactions = 50
probb= 0.04
X = np.random.binomial(number_of_transactions, probb, 20000)
print("Theoretical Mean:", number_of_transactions * probb)
print("Empirical Mean:", np.mean(X))
````
