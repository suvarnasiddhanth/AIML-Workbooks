```python
data = pd.read_csv("/diabetes.csv")
dataset = data
dataset
```

```python
x = dataset.iloc[:, :-1].values
x
```

```python
y = dataset.iloc[:, -1].values
y
```

```python
dataset.iloc[:, 1:6] = dataset.iloc[:, 1:6].replace(0, np.nan)
dataset
```

```python
dataset.isnull().sum()
```

```python
dataset = dataset.drop(columns=['Insulin'])
```

```python
dataset = dataset.drop(columns=['SkinThickness'])
```

```python
dataset.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer = imputer.fit(x[:, 1:4])
x[:, 1:4] = imputer.transform(x[:, 1:4])
x
```

```python
dataset.duplicated().sum()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 45)
```

```python
x_test
```

```python
from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
x_train = sc_x.fit_transform(x_train)
x_test = sc_x.transform(x_test)
```

```python
x_test
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 54)
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test)
y_pred
```

```python
y_test
```

```python
#compare y_pred and y_test
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
Accuracy = (cm[0][0]+cm[1][1])/(cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1])
Accuracy
```

```python
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred)*100, "%")
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```
