```python
store1 = [10.00, 11.00, 12.34, 5.34]
store2 = [19.00, 10.10, 12.34, 5.01]
store3 = [15.00, 12.00, 11.00, 6.00]
cheapest = map(min, store1, store2, store3)
costliest = map(max, store1, store2, store3)
print(list(cheapest))
print(list(costliest))
```

```python
def calculateSquare(num):
    return num * num
numbers = (1, 2, 3, 4)
result = map(calculateSquare, numbers)
print(list(result))
```

```python
nums2 = [1, 2, 3, 4]
result2 = map(lambda x: x * x, nums2)
print(list(result2))
```

```python
numbers1 = [1,2,3,4]
numbers2 = [5,6,7,8]
result3 = map(lambda x, y: x + y, numbers1, numbers2)
print(list(result3))
```

```python
people = ['python programming language', 'Computer Lab Test', 'Vinod Scored A', 'Kumar scored B']
def split_title_and_name(person):
    title = person.split()[0]
    lastname = person.split()[-1]
    return '{} {}'.format(title, lastname)
result4 = map(split_title_and_name, people)
print(list(result4))
```
