19. Use the following dataset:

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)


Perform the following analysis:

- Display the first five records.
- Find the dimensions of the dataset.
- Find the average final mark.
- Find the highest-scoring student.
- Find the lowest-scoring student.
- Display students scoring above 80.
- Find average marks by department.
- Find students having attendance above 80%.
- Sort students by FinalMarks.
- Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail.
- Add a performance category:
	≥90 → Excellent
	≥80 → Very Good
	≥70 → Good
	<70 → Needs Improvement
- Find the department having the highest average FinalMarks.

```python
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)

print("Display the first five records\n", df[:5])
print("Dimensions of the dataset.\n", df.shape)
print("Average final mark: ", df["FinalMarks"].mean())
print("Highest-scoring student:\n", df[(df["FinalMarks"]) == df["FinalMarks"].max()])
print("Lowest-scoring student:\n", df[(df["FinalMarks"]) == df["FinalMarks"].min()])
print("Display students scoring above 80:\n", df[(df["FinalMarks"]) > 80])
print("Average marks by department:n", df.groupby('Department')['FinalMarks'].agg("mean"))

```

```python
print("Students having attendance above 80%:\n", df[(df["Attendance"]) > 80])
print("Sort students by FinalMarks:\n", df.sort_values("FinalMarks"))
print("Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail:")
df["Result"] = np.select(
    [
        df["FinalMarks"] > 70,
    ],
    [
        "Pass"
    ],
    default="Fail"
)
df
```

```python
print("Add a performance category:\n≥90 → Excellent\n≥80 → Very Good\n≥70 → Good\n<70 → Needs Improvement\n")
df["Performance"] = np.select(
    [
        df["FinalMarks"] >= 90,
        df["FinalMarks"] >= 80,
        df["FinalMarks"] >= 70
    ],
    [
        "Excellent",
        "Very Good",
        "Good"
    ],
    default="Needs Improvement"
)
df
```

```python
print("Highest average marks by department:", df.groupby('Department')['FinalMarks'].agg("mean").max())
```

20. Create:

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})


Determine:

- Average salary.
- Maximum and minimum salary.
- Employee receiving the highest salary.
- Employees earning above ₹60,000.
- Average salary for each department.
- Average experience by department.
- Number of employees in each department.
- Sort employees by salary.
- Find IT employees with more than 3 years of experience.
- Add a SalaryLevel column:
	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low

```python
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})
```

```python
print("Average Salary:", employees["Salary"].mean())
print("Maximum Salary:", employees["Salary"].max())
print("Minimum Salary:", employees["Salary"].min())
print("\nHighest Paid Employee:\n", employees.loc[employees["Salary"].idxmax()])
```

```python
print("Employees earning above ₹60,000:\n", employees[employees["Salary"] > 60000])
print("\nAverage Salary by Department:\n", employees.groupby("Department")["Salary"].mean())
print("\nAverage Experience by Department:\n", employees.groupby("Department")["Experience"].mean())
print("\nNumber of Employees in Each Department:\n", employees["Department"].value_counts())
```

```python
print("\nEmployees Sorted by Salary:\n", employees.sort_values("Salary"))
print("\nIT Employees with More Than 3 Years Experience:\n", employees[(employees["Department"] == "IT") & (employees["Experience"] > 3)])
```

```python
"""Add a SalaryLevel column:
	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low"""

employees["SalaryLevel"] = np.select(
    [
        employees["Salary"] >= 75000,
        employees["Salary"] >= 50000
    ],
    [
        "High",
        "Medium"
    ],
    default="Low"
)
employees
```
