```python
marks = np.array([[12,23,41],[46,24,73],[75,37,74],[23,44,44]])
marks
```

```python
student_total = np.sum(marks, axis=1)
print("Total of each student:", student_total)
```

```python
meanstud = np.mean(marks)
print("Mean of each student:", meanstud)
```

```python
argmaxstud = np.argmax(student_total)
print("Max of each student:", argmaxstud)
```
