```python
#10. The following array contains marks of four students in three subjects:

marks = np.array([
    [80, 95, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

#Calculate:

#Total marks of each student
#Average marks of each student
#Average marks in each subject
#Student with the highest total
#Highest mark in each subject.

#Total marks of each student
student_total = np.sum(marks, axis=1)
print("Total of each student:", student_total)



#Average marks of each student
meanstud = np.mean(marks, axis=1)
print("Mean of each student:", meanstud)

#Average marks in each subject
meansub = np.mean(marks, axis=0)
print("Mean of each student:", meansub)

#Student with the highest total
argmaxstud = np.argmax(student_total)
print("Max student:", argmaxstud)

#Highest mark in each subject.
argmaxstud = np.argmax(marks, axis=0)
print("Max subject:", argmaxstud)
```

```python
#12. Daily temperatures for two weeks are:

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

#Find:
#Mean temperature
print("Mean temperature:", temperature.mean())

#Maximum and minimum temperatures
print("Maximum temperatures:",temperature.max())
print("Minimum temperatures:",temperature.min())

#Number of days above the average temperature
print("Number of days above the average temperature: ",temperature[temperature>=temperature.mean()])

#Temperatures greater than 34°C
print("Temperatures greater than 34°C:", temperature[temperature>34])

#Difference between maximum and minimum
print("Difference between maximum and minimum:", temperature.max()-temperature.min())

#Standard deviation
print("Standard Deviation:", temperature.std())
```

```python
#13. Monthly sales of a company are:
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

#Determine:

import calendar

#Total annual sales
print("Total annual sales:", sales.sum())

#Average monthly sales
print("Average annual sales:", sales.mean())

#Month with the highest sales
print("Month with the highest sales:", calendar.month_name[np.argmax(sales) + 1])

#Month with the lowest sales
print("Month with the highest sales:", calendar.month_name[np.argmin(sales) + 1])

#Number of months with sales above ₹20,000
print("Number of months with sales above ₹20,000:", sales[sales>20000].size)

#Percentage increase from January to December
print("Percentage increase from January to December:", (sales.max()/sales.min())*100)

#Hint: Remember that NumPy indexing starts at 0, while month numbering normally starts at 1.
```

```python
#14. Consider marks of five students in four subjects:

marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

#Write a NumPy program that calculates:

#Total marks of every student
print("Total marks of every student:", marks.max(1))

#Average of every student
print("Average of every student:", marks.mean(1))

#Average mark for every subject
print("Average of every subject:", marks.mean(0))

#Highest-scoring student
print("Highest-scoring student index:", np.argmax(marks.sum(1)))

#Lowest-scoring student
print("Lowest-scoring student index:", np.argmin(marks.sum(1)))

#Students whose average is greater than 80
avgmarks=marks.mean(1)
print("Students whose average is greater than 80:", np.where(marks.mean(1) >= 80)[0])

#Overall class average
print("Overall class average:", marks.mean())

#Then assign a performance category based on average:

#|  Average | Category          |
#| -------: | ----------------- |
#|     ≥ 85 | Excellent         |
#| 70–84.99 | Good              |
#| 50–69.99 | Average           |
#|     < 50 | Needs Improvement |

#Challenge: Try assigning these categories using np.where() or np.select() instead of a Python loop.
print("\nOverall class catergories:")
allavg=marks.mean(1)
category=categories = np.select(
    [
        allavg >= 85,
        allavg >= 70,
        allavg >= 50
    ],
    [
        "Excellent",
        "Good",
        "Average"
    ],
    default="Needs Improvement"
)
print(category)

#tabular form
print("\n\tAverage | Category\n",'-' * 40)
for i in range(marks.shape[0]):
    print(f"\t{allavg[i]:.2f}   |  {category[i]}")
```

```python
#23. Create a NumPy array containing 100 random exam marks and determine how many students fall into the ranges 0–39, 40–59, 60–79, and 80–100.
randmarks = np.random.randint(0,101,100)
markrange=np.select(
    [randmarks >=80,
     randmarks >= 60,
     randmarks >= 40
    ],
    [
        "80-100",
        "60-79",
        "40-59"
    ],
    default="0-39"
)
vals,counts = np.unique(markrange, return_counts=True)
print(dict(zip(vals,counts)))
```
