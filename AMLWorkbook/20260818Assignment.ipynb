```python
marks_1 = pd.Series([78,85,92,67,88])
print("Complete Series:")
print(marks_1)
print("Values:", marks_1.values)
print("Index:",marks_1.index)
print("Data type:", marks_1.dtype)
print("Mean:", marks_1.mean())
print("Maximum:", marks_1.max())
print("Minimum:", marks_1.min())
```
