```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("drive/My Drive/Iris.csv")
dataset = data
dataset
```

```python
dataset.info()
```

```python
x = dataset.iloc[:, 1:-1].values
x
```

```python
dataset.describe()
```

```python
y = dataset.iloc[:, -1].values
y
```

```python
dataset.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 45)
```

```python
from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
x_train = sc_x.fit_transform(x_train)
x_test = sc_x.transform(x_test)
```

```python
x_test
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 45)
#classifier = DecisionTreeClassifier(criterion = 'gini', random_state = 45)
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test)
y_pred
```

```python
#compare y_pred and y_test
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
Accuracy = (cm[0][0]+cm[1][1]+cm[2][2])/(cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1]+cm[2][0]+cm[2][1]+cm[2][2])*100
Accuracy
```

```python
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred)*100, "%")
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
new_individual = np.array([[5, 3.5, 1.5, 0.2]])
new_individual_scaled = sc_x.transform(new_individual)
new_individual_scaled
```

```python
prediction = classifier.predict(new_individual_scaled)
if prediction == 1:
  print("1st species")
elif prediction == 2:
  print("2nd species")
else :
  print("3rd species")
```
